```python
import weaviate
from weaviate.classes.init import Auth
from weaviate.classes.config import Property, DataType, Configure

client = weaviate.connect_to_weaviate_cloud(
    cluster_url=WEAVIATE_URL,
    auth_credentials=Auth.api_key(WEAVIATE_API_KEY),
    headers={"X-OpenAI-Api-Key": OPENAI_API_KEY}
)


# Delete existing collection if it exists
if client.collections.exists("Chatbot"):
    client.collections.delete(name="Chatbot")

# Create new collection
collection = client.collections.create(
    name="Chatbot",
    description="Documents for chatbot",
    vector_config=Configure.Vectors.text2vec_openai(model="ada"),
    properties=[
        Property(
            name="content",
            data_type=DataType.TEXT,
            description="The content of the paragraph",
            skip_vectorization=False,
            vectorize_property_name=False
        )
    ]
)

from langchain_community.vectorstores.weaviate import Weaviate as LCWeaviate

from langchain_weaviate import WeaviateVectorStore
from langchain.embeddings import OpenAIEmbeddings

embeddings = OpenAIEmbeddings(model="text-embedding-3-small", openai_api_key=OPENAI_API_KEY)

vs = WeaviateVectorStore(
    client=client,
    index_name="Chatbot",
    text_key="content",
    attributes=["source"],
    embedding=embeddings
)

```

```python
# load text into the vectorstore
text_meta_pair = [(doc.page_content, doc.metadata) for doc in docs]
texts, meta = list(zip(*text_meta_pair))
vs.add_texts(texts, meta)
```

## Similarity Measurement

```python
query = "what is a yolo?"

# retrieve text related to the query
docs = vs.similarity_search(query, top_k=20)
```

```python
docs
```

```python
from langchain.chains.question_answering import load_qa_chain
from langchain.llms import OpenAI
```

```python
# define chain
chain = load_qa_chain(
    OpenAI(openai_api_key = OPENAI_API_KEY,temperature=0),
    chain_type="stuff")
```

```python
# create answer
chain.run(input_documents=docs, question=query)
```
